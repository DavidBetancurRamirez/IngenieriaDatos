# Datos nulos

## Diferentes formas que se ven los nulos

### Como se ven datos nulos en Python

```python
print(None or True,
      None or False,
      None or 0,
      None or 1,
      None or [],
      None or [1, 2, 3],
      None or '',
      None or 'Hello',
      None or {},
      None or {'a': 1, 'b': 2},
      type(None),
      sep = '\n'
)
```

### Como se manejan datos nulos en Numpy

```python
print(np.nan or True,
      np.nan == np.nan,
      type(np.nan),
      sep = '\n'
)
```

### Como se manejan los datos nulos en Pandas

```python
prueba_faltante = pd.DataFrame.from_dict(data = dict(
  x = [0, 1, np.nan, np.nan, None],
  y = [0, 1, pd.NA, np.nan, None]
))

prueba_faltante
```

## Metodos detectar datos nulos

### Detectar datos nulos desde un Dataframe

Metodo .isna() de pandas devuelve un DataFrame de True/False dependiendo si identifica en la entrada un registro nulo

```python
prueba_faltante.isna()
```

```python
print("Cantidad de valores Nulos", prueba_faltante.missing.number_missing())
print("Cantidad de valores NO Nulos", prueba_faltante.missing.number_complete())
```

### Detectar datos nulos en Series de tiempo

```python
pd.Series([1, np.nan])
```

```python
pd.Series([pd.to_datetime('2022-01-01'), np.nan])
```

## Extensión de Pandas de Datos Nulos

### **iloc**
Llena con el valor que se de en los valores que se pasan.
[desde:hasta+1, index columna]

```python
df = pd.DataFrame.from_dict(data = {"a": list('asdfasdfas'), "b": range(0,10)})
df.iloc[2:5, 0] = None
df.iloc[6:7, 1] = None
df
```

```python
print("Cantidad de valores Nulos", df.missing.number_missing())
print("Cantidad de valores NO Nulos", df.missing.number_complete())
```
